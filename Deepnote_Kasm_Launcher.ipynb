# Deepnote KasmVNC Launcher

Run the first code cell inside Deepnote. You do not need the Terminal.

The start cell runs in detached mode: it can finish, and the KasmVNC desktop should keep running in the background. When it says **KasmVNC desktop is ready**, open your Deepnote incoming-connections URL and log in with the username/password printed by the cell.

Run the stop cell when you want to shut the desktop down.


```python
%%bash
set -Eeuo pipefail
cd ~/work 2>/dev/null || cd ~

# If this notebook was imported at the project root and scripts/ exists, use it.
# Otherwise, replace any old os folder with the correct Arena branch.
if [ -x scripts/deepnote-desktop.sh ]; then
  bash scripts/deepnote-stop.sh || true
  KASMVNC_DETACH=true bash scripts/deepnote-desktop.sh
else
  if [ -d os ]; then
    bash os/scripts/deepnote-stop.sh || true
  fi
  rm -rf os
  git clone --branch arena/01a0eb07-os --single-branch https://github.com/Joercat/os.git os
  cd os
  KASMVNC_DETACH=true bash scripts/deepnote-desktop.sh
fi
```

## Stop the desktop

Run this cell when you want to stop KasmVNC/XFCE.


```python
%%bash
set -Eeuo pipefail
cd ~/work/os 2>/dev/null || cd ~/os
bash scripts/deepnote-stop.sh
```
